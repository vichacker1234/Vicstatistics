## **Cargar el modelo de bot a  usar**

```python
%%writefile app/data_processing.py

import pandas as pd

def melt_weeks(df):

    week_cols = [c for c in df.columns if "W" in c]

    df_long = df.melt(
        id_vars=["COUNTRY","CITY","ZONE","METRIC"],
        value_vars=week_cols,
        var_name="WEEK",
        value_name="VALUE"
    )

    return df_long
```
